```python
# 01_collecte.ipynb
# Data Collection

import pandas as pd
import numpy as np
import yfinance as yf
import requests
import matplotlib.pyplot as plt

print("Environment ready")
```

```python
ticker = "^GSPC"   # S&P 500

data = yf.download(
    ticker,
    start="2012-01-01",
    end="2024-12-31",
    auto_adjust=False,
    progress=False
)

data.head()
```

```python
data.columns = data.columns.droplevel(1)

data.head()
```

```python
data.to_csv("../data/raw/sp500_prices.csv")

print("S&P 500 saved successfully")
```

```python
from pytrends.request import TrendReq
pytrends = TrendReq(
    hl='en-US',
    tz=360
)

print("Google Trends connected")
```

```python
keywords = ["Artificial Intelligence"]

pytrends.build_payload(
    keywords,
    timeframe="2012-01-01 2024-12-31",
    geo=""
)

ai_trends = pytrends.interest_over_time()

ai_trends.head()
```

```python
ai_trends = ai_trends.drop(columns=["isPartial"])

ai_trends.to_csv("../data/raw/google_trends_ai.csv")

print("Google Trends AI saved successfully")
```

```python
from pandas_datareader import data as pdr
inflation = pdr.DataReader(
    "CPIAUCSL", #Consumer Price Index for All Urban Consumers
    "fred",
    "2012-01-01",
    "2024-12-31"
)

inflation.head()
```

```python
inflation["Inflation"] = inflation["CPIAUCSL"].pct_change() * 100

inflation.head()
```

```python
inflation = inflation[["Inflation"]]

inflation.to_csv("../data/raw/inflation.csv")

print("Inflation saved successfully")
```
