```python
import pandas as pd
import numpy as np
import os
sp500 = pd.read_csv(
    "../data/raw/sp500_prices.csv",
    index_col="Date",
    parse_dates=True
)

ai_news = pd.read_csv(
    "../data/raw/google_trends_ai.csv",
    index_col="date",
    parse_dates=True
)

vix = pd.read_csv(
    "../data/raw/vix.csv",
    index_col="Date",
    parse_dates=True
)

inflation = pd.read_csv(
    "../data/raw/inflation.csv",
    index_col="DATE",
    parse_dates=True
)

print("All datasets loaded")

```

```python
print(sp500.head())
print(ai_news.head())
print(vix.head())
print(inflation.head())
```

```python
sp500 = sp500[["Close"]].copy()

sp500.columns = ["SP500_Close"]

sp500.head()
```

```python
sp500_monthly = sp500.resample("ME").last()

sp500_monthly.head()
```

```python
vix_monthly = vix.resample("ME").mean()

vix_monthly.head()
```

```python
sp500_monthly = sp500_monthly.rename(
    columns={"Close": "SP500_Close"}
)

vix_monthly = vix_monthly.rename(
    columns={"Close": "VIX"}
)

ai_news = ai_news.rename(
    columns={"Artificial Intelligence": "AI_News"}
)
```

```python
ai_news.index = pd.to_datetime(ai_news.index) + pd.offsets.MonthEnd(0)

ai_news.head()
```

```python
inflation.index = pd.to_datetime(inflation.index) + pd.offsets.MonthEnd(0)

inflation.head()
```

```python
vix = vix[["Close"]].copy()

vix.columns = ["VIX"]

vix_monthly = vix.resample("ME").last()

vix_monthly.head()
```

```python
print(sp500_monthly.index[:5])
print(ai_news.index[:5])
print(vix_monthly.index[:5])
print(inflation.index[:5])
```

```python
dataset = (
    sp500_monthly
    .join(ai_news, how="inner")
    .join(vix_monthly, how="inner")
    .join(inflation, how="inner")
)

dataset.head()
```

```python
dataset.to_csv("../data/clean/merged_dataset.csv")

print("Dataset saved")
```

```python
dataset["SP500_Return"] = dataset["SP500_Close"].pct_change()

dataset.head()
```

```python
dataset["SP500_Volatility"] = (
    dataset["SP500_Return"]
    .rolling(window=3)
    .std()
)

dataset.head()
```

```python
dataset.isna().sum()
```

```python

dataset["AI_News_Change"] = dataset["AI_News"].pct_change() #month-to-month change in AI interest.
dataset.head()
```

```python
dataset["VIX_Change"] = dataset["VIX"].pct_change()

dataset.head()
```

```python
dataset["SP500_Return_Lag1"] = dataset["SP500_Return"].shift(1)

dataset["VIX_Lag1"] = dataset["VIX"].shift(1)

dataset["AI_News_Lag1"] = dataset["AI_News"].shift(1)

dataset.head()
```
