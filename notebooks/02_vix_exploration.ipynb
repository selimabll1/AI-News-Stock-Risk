```python

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Locate repository root
ROOT = Path.cwd()

if not (ROOT / "data").exists():
    ROOT = ROOT.parent

# Load processed VIX data
vix = pd.read_csv(
    ROOT / "data/processed/vix_monthly.csv",
    parse_dates=["DATE"]
)

# Inspect data
display(vix.head())
print("Observations:", len(vix))
print(vix["VIX"].describe())

# Plot the monthly VIX
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    vix["DATE"],
    vix["VIX"],
    linewidth=1.5,
    label="Monthly VIX"
)

ax.set_title(
    "CBOE VIX — Monthly Average (2012–2024)"
)
ax.set_xlabel("Date")
ax.set_ylabel("VIX Index")
ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()

# Save the research figure
output = ROOT / "reports/figures/vix_monthly.png"
output.parent.mkdir(parents=True, exist_ok=True)

fig.savefig(output, dpi=300)
plt.show()
```
